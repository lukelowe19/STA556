Part 5 - a function that changes its input

```python
def add_value(values):
    values.append(100)

data = [1, 2, 3]
add_value(data)

print(data)
```

```python
def add_value(values):
    result = values.copy()
    result.append(100)
    return result

data = [1, 2, 3]
new_data = add_value(data)

print(data)
print(new_data)
```

I would prefer the non-mutating version, as it preserves the original dataset, and does not override the origial list. 

Part 7 - Lists

```python
scores = [72, 81, 91, 68, 88, 95, 77, 84]
print(scores[0])
print(scores[-1])
print(scores[2:5])
print(len(scores))
scores.append(90)
print(scores)
scores.sort()
print(scores)
scores.reverse()
print(scores)
```

```python
print(f"Number of Observations: {len(scores)}"),
print(f"Minimum score: {min(scores)}"),
print(f"Maximum score: {max(scores)}"),
print(f"Mean Score: {(sum(scores))/len(scores)}")
```

Part 10 — Sets

```python
states = [
    "Arizona", "Arizona", "Utah",
    "Nevada", "Arizona", "Utah"
]

unique_states = set(states)
```

```python
unique_states
"Arizona" in unique_states
"California" in unique_states
print(unique_states)
```

Because a set doesn't allow for duplicates, it can organize the data by unique values rather than including repition.
