Part 21: Mini analysis challenge

```python
participants = pd.DataFrame(
    {
        "id": range(1001, 1011),
        "age": [22, 34, 29, 41, 19, 37, 26, 45, 31, 28],
        "group": ["A", "B", "A", "C", "B", "A", "C", "B", "A", "C"],
        "score": [78, 91, 85, 88, 72, 95, 81, 67, 90, 84]
    }
).set_index("id")
```

```python
# aged 30 and older
over_thirty = participants.loc[
(participants["age"]>= 30)
]
over_thirty
```

```python
# Groups A or C
groupAorC = participants.loc[
    (participants["group"] == "A") |
    (participants["group"] == "C")
]
groupAorC
```

```python
score_least_85 = participants.loc[
    (participants["score"] >= 85)
]
score_least_85
```

```python
thirty_least85 = participants.loc[
    (participants["age"]>= 30) &
    (participants["score"]>= 85)
]
thirty_least85
```

```python
high_scores = (
    (participants["score"] >= 90)
)

high_score_summary = participants.loc[
    high_scores,
    ["group","score"]
]
high_score_summary
```

```python
not_group_b = participants.loc[
    participants["group"] != "B" 
]
not_group_b
```

Function Challenge

```python
def select_minimum_score(
        df: pd.DataFrame,
        minimum_score: float
) -> pd.DataFrame:
    df_filtered = df.loc[df["score"]> minimum_score]
    return df_filtered

select_minimum_score(participants,80)

```
